# Repeated-measures rainclouds on the AnxietyPaper data

This section demonstrates `paired_raincloud`, the dedicated function for paired /
within-subject data. It is self-contained: it loads the Bristol AnxietyPaper sheet
(the same dataset used elsewhere in this tutorial) and melts it to long form.

```python
import pandas as pd

# Same source + melt used elsewhere in the tutorial
df = pd.read_csv(
    "https://data.bris.ac.uk/datasets/112g2vkxomjoo1l26vjmvnlexj/"
    "2016.08.14_AnxietyPaper_Data%20Sheet.csv",
    sep=",",
)
ddf = pd.melt(
    df,
    id_vars=["Participant"],
    value_vars=["AngerUH", "DisgustUH", "FearUH", "HappyUH"],
    var_name="EmotionCondition",
    value_name="Sensitivity",
)
ddf.head()
```

## Connecting paired observations

Every `Participant` appears across all four emotion conditions, so this is
repeated-measures data. `paired_raincloud` draws a line through each
participant's points, making within-subject change visible across conditions.
It takes the same arguments as `RainCloud`, plus an `id` column.

```python
import matplotlib.pyplot as plt

import ptitprince as pt

order = ["AngerUH", "DisgustUH", "FearUH", "HappyUH"]

f, ax = plt.subplots(figsize=(9, 5))
pt.paired_raincloud(
    x="EmotionCondition",
    y="Sensitivity",
    data=ddf,
    id="Participant",  # <- connect each participant across conditions
    order=order,
    orient="h",
    move=0.2,  # shift the rain aside so the lines are readable
    line_alpha=0.25,
    point_size=3,
    ax=ax,
)
ax.set_title("Within-participant sensitivity across emotion conditions")
plt.show()
```

Each thin line is one participant. Compared with the standard raincloud, the
paired view also shows *how individuals move* between conditions — not just how
the group distributions differ. Missing measurements (`NaN`) break that
participant's line. `hue`/`dodge` are not supported by `paired_raincloud` (they
raise an error, since the point-to-subject mapping is ambiguous with dodged
sub-groups — use `RainCloud` for those plots).

```python
# Vertical orientation works too:
f, ax = plt.subplots(figsize=(7, 6))
pt.paired_raincloud(
    x="EmotionCondition",
    y="Sensitivity",
    data=ddf,
    id="Participant",
    order=order,
    orient="v",
    move=0.2,
    line_alpha=0.25,
    point_size=3,
    ax=ax,
)
plt.show()
```
